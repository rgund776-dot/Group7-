```python
class PatientExam:
    
    def __init__(self, exam_id: int, date: str, name: str, weight: int, height: float):
        self.exam_id = exam_id
        self.date = date 
        self.name = name
        self.weight = weight 
        self.height = height 
        
    def get_bmi(self) -> float:
        return self.weight / (self.height ** 2) 

    def get_exam_month(self) -> int: 
        return int(self.date.split("/")[0])
```

```python
# create an empty list to store the objects
patients = []

#open our dataset 
with open("patient_data.csv", "r") as file:
    reader = csv.reader(file)
    
    next(reader) #skip the header
    #Create a patientexam object for every row 
    for row in reader:
        exam_id = int(row[0])
        date = row[1]
        name = row[2]
        weight = int(row[3])
        height = float(row[4])
        
        patient = PatientExam(exam_id, date, name, weight, height)
        patients.append(patient)

# Check how many objects were created
print("Number of patient exams:", len(patients))
```

```python
# check the first PatientExam object created from the CSV 

first_patient = patients[0]

print("Name:", first_patient.name)
print("BMI:", first_patient.get_bmi())
print("Exam month:", first_patient.get_exam_month())
```

```python
# add the BMI of every patient together 
total_bmi = 0 

for patient in patients: 
    total_bmi += patient.get_bmi()
# divide the total BMI by the number of patients
average_bmi = total_bmi / len(patients)

print("Average BMI:", average_bmi)

# create 12 counters, one for each month
month_counts = [0] * 12 

#count how many exams occurred in every month 
for patient in patients: 
    month = patient.get_exam_month()
    month_counts[month - 1] += 1 

#find the momth with the most exams
busiest_month = month_counts.index(max(month_counts)) + 1 

print("Busiest Month:", busiest_month)
```
