```python
assert len(data) == 20 
assert data.dtype.names == ('Date', 'Open', 'Close', 'Volume') 
assert len(date) == 20 
assert len(openp) == 20 
assert len(closep) == 20 
assert len(vol) == 20 
assert np.issubdtype(openp.dtype, np.number)
assert np.issubdtype(closep.dtype, np.number)
assert np.issubdtype(vol.dtype, np.number)
print("PASSED CSV LOADING AND EXTRACTION")
```

```python
# Calculate daily returns
rate = lab.calculate_returns(date, openp, closep)
expected_rate = (closep - openp) / openp * 100 
assert np.allclose(rate, expected_rate)
assert len(rate) == 20
print("Expected first five", expected_rate[:5])
print("Actual first five", rate[:5])
print("PASSED: Daily return calculations")
```

```python
# Calculate statistics
mean_rate, median_rate, std_rate = lab.calculate_statistics(rate)
assert np.isclose(mean_rate, np.mean(rate))
assert np.isclose(median_rate, np.median(rate))
assert np.isclose(std_rate, np.std(rate))
print("PASSED: MEAN, MEDIAN AND STD")
```
