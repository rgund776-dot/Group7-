```python
#task 3, sorting the list into acsending order 
import csv 

teams = []

with open("f1_points.csv", "r") as file:
    reader = csv.DictReader(file)
    
    for row in reader:
        
        driver = Driver(row["Driver"], int(row["Points"]))
        
        team = None 
        
        for existing_team in teams:
            if existing_team.name == row["Team"]:
                team = existing_team
                break 
                
        if team is None:
            team = Team(row["Team"])
            teams.append(team)
        
        team.add_driver(driver)
        
        
sorted_teams = sorted(teams)

for team in sorted_teams:
    print(team)
```
