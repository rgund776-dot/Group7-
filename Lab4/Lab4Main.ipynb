```python
#read the file and get the number of teams 
from classes1 import Driver, Team
import csv 

teams = []

with open("f1_points.csv", "r") as file:
    reader = csv.DictReader(file)
    
    next(reader)
    
    for row in reader:
        
        driver = Driver(row["Driver"], int(row["Points"]))
        
        team = None 
        
        for existing_team in teams:
            if existing_team.name == row["Team"]:
                team = existing_team
                break 
                
        if team is None:
            team = Team(row["Team"])
            teams.append(team)
        
        team.add_driver(driver)
        
print("Number of Teams", len(teams))
```

```python
# Task 3: Sorting the teams into ascending order
from classes1 import Driver, Team
import csv

teams = {}

with open("f1_points.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:

        driver = Driver(row["Driver"], int(row["Points"]))

        if row["Team"] not in teams:
            teams[row["Team"]] = Team(row["Team"])

        team = teams[row["Team"]]
        team.add_driver(driver)

# Sort the Team objects into ascending order
sorted_teams = sorted(teams.values())

for team in sorted_teams:
    print(team)
    print()
```
