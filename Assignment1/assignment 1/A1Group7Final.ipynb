```python
#format dection to make sure we use the correct function from file.IO 
def load_data(filename):
    with open(filename, 'r') as file:
        first_line = file.readline()
    
    if '<' in first_line:
        return load_from_html(filename)
    elif ',' in first_line:
        return load_from_csv(filename)
    else: 
        raise Exception('Error, data must be in valid CSV or HTML format')
        
html_table = load_data('./student_dataset.txt')
csv_table = load_data('./census_test.txt')

print("HTML rows:", len(html_table))
print("CSV rows:", len(csv_table))
```
