```python
import csv 
from exceptions import TextFormatException, MissingValueException, MeasurementUnitException


import csv
from exceptions import TextFormatException, MissingValueException, MeasurementUnitException


def parse_row(row):
    if len(row) != 5:
        raise TextFormatException("Wrong number of values")

    if "" in row:
        raise MissingValueException("A value is missing")

    try:
        exam_id = int(row[0])
        weight = float(row[3])
        height = float(row[4])
    except ValueError:
        raise TextFormatException("Invalid number format")

    if height > 3:
        raise MeasurementUnitException("Height is in feet")

    return [exam_id, row[1], row[2], weight, height]


def calculate_bmi(weight, height):
    return weight / (height ** 2)


def main():
    with open("data.csv", "r") as file:
        reader = csv.reader(file)
        next(reader)

        with open("bmi_output.csv", "w") as output_file:
            writer = csv.writer(output_file)
            writer.writerow(["Exam ID", "BMI"])

            for row in reader:
                try:
                    exam_id, date, name, weight, height = parse_row(row)

                    bmi = calculate_bmi(weight, height)

                    writer.writerow([exam_id, f"{bmi:.2f}"])

                except TextFormatException as e:
                    print(f"Exam ID {row[0]}: {e}")

                except MissingValueException as e:
                    print(f"Exam ID {row[0]}: {e}")

                except MeasurementUnitException as e:
                    print(f"Exam ID {row[0]}: {e}")


main()
```

the goal is to process the patient data and find BMI for all the records that are VALID 
there are 3 types of data entry errors: 
1. Missing values 
2. Heights entered in feet instead of metres 
3. rows with wrong number of values

rows with errors aren't put on the output file, and we had the function print which exams have an error.
