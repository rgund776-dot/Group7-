```python
from classes1 import Team, Driver
import csv
```

```python
teams = {}
with open("f1_points.csv", "r") as file:
	reader = csv.reader(file)

	#skip header
	next(reader)

	for row in reader:
		driver_name = row[0]
		team_name = row[1]
		points = int(row[2])

		if team_name not in teams:
			teams[team_name] = Team(team_name)

		driver = Driver(driver_name, points)
		teams[team_name].add_driver(driver)
	
	print("Number of Teams:", len(teams))
```

```python
red_bull = teams["RED BULL RACING HONDA RBPT"]

print("Team:", red_bull.name)
print("Drivers:", red_bull.drivers)
```
