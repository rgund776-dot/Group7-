# Test A Valid Melee Attack 

```python
fighter = Fighter("Aragorn", 100, 8)
target = Mage("Enemy Mage", 80, 5)
sword = MeleeWeapon("Sword", 15)

health_before = target.health
attack(fighter, sword, target, distance = 0)

print("Expected target health", 65)
print("Actual Target health", target.health)

assert target.health == 65
print("PASS: Valid Melee Attack")
```

The cell above is testing that if a sword which has no ranged attribute, it will successfully hit the target for damage at a distance of 0. we used the assert function to check that the target health was 65 at the end of the attack

# Test B Valid Ranged Attack 


```python
bow = RangedWeapon("Bow",12,10)
target = Mage("Enemy Mage",80,5)

health_before = target.health 
attack(fighter, bow, target, distance = 5)

print("Health before attack", health_before)
print("Expected health after attack", 68)
print("Actual helath after attack", target.health)
print("Damage received:",health_before - target.health)

assert target.health == 68 
print("PASS: Valid Ranged Attack")
```

The cell above is testing that if a ranged weapon which has a ranged attribute of 10, it will successfully hit the target for damage at a distance of 5 when attacking (5 < 10 so it hits). we used the assert function to check that the target health was 68 at the end of the ranged attack

# Test C Valid Magic Attack


```python
staff = MagicWeapon("Magic staff",20, 10, 5)
mage = Mage("Gandalf", 80, 5)
target = Fighter("Enemy Fighter", 100, 8)

attack(mage, staff, target, distance = 5)

print("Expected target health", 80)
print("Actual target helath", target.health)

assert target.health == 80
print("PASS Valid Magic Attack")
```

The cell above is testing that if a magic weapon which has a ranged attribute of 10, it will successfully hit the target for damage at a distance of 5 when attacking (5 < 10 so it hits). the staff has a mana use value of 5, and the mage character has 5 mana so the attack goes through. we used the assert function to check that the target health was 80 at the end of the ranged attack
