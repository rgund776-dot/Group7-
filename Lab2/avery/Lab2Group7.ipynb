```python
def read_stopwords(filename):
    with open(filename, "r") as file:
        stopwords = set()

        for line in file:
            word = line.strip().lower()

            if word != "":
                stopwords.add(word)

    return stopwords


def read_words(filename, stopwords):
    with open(filename, "r") as file:
        text = file.read().lower()

    punctuation = ".,?!;:'\"()-"

    for character in punctuation:
        text = text.replace(character, "")

    words = text.split()
    filtered_words = []

    for word in words:
        if word not in stopwords:
            filtered_words.append(word)

    return filtered_words


def calculate_term_frequency(word_counts, total_words):
    term_frequency = {}

    for word in word_counts:
        term_frequency[word] = word_counts[word] / total_words

    return term_frequency


# Read the stopwords
stopwords = read_stopwords("stopwords.txt")

# Read the two documents
wordsA = read_words("textA.txt", stopwords)
wordsB = read_words("textB.txt", stopwords)


# Count the words in Document A
word_countsA = {}

for word in wordsA:
    if word in word_countsA:
        word_countsA[word] += 1
    else:
        word_countsA[word] = 1


# Count the words in Document B
word_countsB = {}

for word in wordsB:
    if word in word_countsB:
        word_countsB[word] += 1
    else:
        word_countsB[word] = 1


# Calculate term frequencies
term_frequencyA = calculate_term_frequency(word_countsA, len(wordsA))
term_frequencyB = calculate_term_frequency(word_countsB, len(wordsB))


# Print number of words
print("Number of words in text A:", len(wordsA))
print("Number of words in text B:", len(wordsB))


# Print word counts
print("Word counts for text A:", word_countsA)
print("Word counts for text B:", word_countsB)


# Print term frequencies
print("Term frequencies for text A:", term_frequencyA)
print("Term frequencies for text B:", term_frequencyB)


# Find words common to both documents
common_words = set()

for word in term_frequencyA:
    if word in term_frequencyB:
        common_words.add(word)

print("Common words:", common_words)


# Remove common words from both term frequency dictionaries
for word in common_words:
    del term_frequencyA[word]
    del term_frequencyB[word]


# Print the remaining unique words
print("Unique words in text A:", term_frequencyA)
print("Unique words in text B:", term_frequencyB)
```

```python
# Test 1: Normal case

stopwords = read_stopwords("Stopwords4T1.txt")

wordsA = read_words("textA4T1.txt", stopwords)
wordsB = read_words("textB4T1.txt", stopwords)

word_countsA = {}

for word in wordsA:
    if word in word_countsA:
        word_countsA[word] += 1
    else:
        word_countsA[word] = 1

word_countsB = {}

for word in wordsB:
    if word in word_countsB:
        word_countsB[word] += 1
    else:
        word_countsB[word] = 1

term_frequencyA = calculate_term_frequency(word_countsA, len(wordsA))
term_frequencyB = calculate_term_frequency(word_countsB, len(wordsB))

common_words = set()

for word in word_countsA:
    if word in word_countsB:
        common_words.add(word)

unique_wordsA = {}

for word in word_countsA:
    if word not in common_words:
        unique_wordsA[word] = term_frequencyA[word]

unique_wordsB = {}

for word in word_countsB:
    if word not in common_words:
        unique_wordsB[word] = term_frequencyB[word]

print("Remaining Words in Document 1")
print(wordsA)

print("Remaining Words in Document 2")
print(wordsB)

print("Term Frequencies for Document 1")
print(term_frequencyA)

print("Term Frequencies for Document 2")
print(term_frequencyB)

print("Common words:")
print(common_words)

print("Unique words in Document 1:")
print(unique_wordsA)

print("Unique words in Document 2:")
print(unique_wordsB)
```
