```python
from data_processing import print_stats 
from file_IO import load_from_csv


small_table = []

with open('./census_dataset.txt', 'r') as file: 
    columns = file.readline().strip().split(',')
    
    for i in range(10):
        values = file.readline().strip().split(',')
        
        row = {}
        
        for j in range(len(columns)):
            value = values[j]
            try:
                value = float(value)
            except ValueError:
                pass 
            
            row[columns[j]] = value 
        
        small_table.append(row)
        
print_stats(small_table)
```
