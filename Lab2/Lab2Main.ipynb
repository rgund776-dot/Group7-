```python
def read_stopwords(filename):
    with open(filename, "r") as file:
        stopwords = set()
        for line in file:
            word = line.strip().lower()
            stopwords.add(word)
            
    return stopwords 


def read_words(filename, stopwords):
    with open(filename, "r") as file: 
        text = file.read().lower()
        
    punctuation = ".,?!;:'\"()-" 
    for character in punctuation: 
        text = text.replace(character, "")
    
    words = text.split()
    filtered_words = [] 
    
    for word in words: 
        if word not in stopwords:
            filtered_words.append(word)
    
    return filtered_words 

def count_words(words):
    word_counts = {}
    for word in words:
        if word in word_counts:
            word_counts[word] += 1 
        else: 
            word_counts[word] = 1 
            
    return word_counts 

def calculate_term_frequency(word_counts, total_words):
    term_frequency = {}
    for word in word_counts:
        term_frequency[word] = word_counts[word] / total_words
        
    return term_frequency 

def find_common_words(dictionary1,dictionary2):
    common_words = set(dictionary1.keys()) & set(dictionary2.keys())
    
    return common_words 

def remove_common_words(dictionary1,dictionary2, common_words):
    for word in common_words:
        del dictionary1[word]
        del dictionary2[word]
        
        
def main(): 
    #read stopwords 
    stopwords = read_stopwords("stopwords.txt")
    #read both text files 
    wordsA = read_words("textA.txt", stopwords)
    wordsB = read_words("textB.txt", stopwords)
    
    print("Number of words in text A", len(wordsA))
    print("Number of words in text B",len(wordsB))
    
    #count up how many words 
    countsA = count_words(wordsA)
    countsB = count_words(wordsB)
    
    print("\n Word counts for text A:")
    print(countsA)
    print("\n Word counts for text B:")
    print(countsB)
    
    #find term frequencies 
    ftA = calculate_term_frequency(countsA, len(wordsA))
    ftB = calculate_term_frequency(countsB, len(wordsB))
    print("\n Term frequencies for text A")
    print(ftA)
    print("\n Term frequencies for text B")
    print(ftB)
    
    #find common words in docs 
    common_words = find_common_words(countsA, countsB)
    print("\n Words common to both documents")
    print(common_words)
    
    #remove the common ones 
    remove_common_words(ftA,ftB,common_words)
    print("\nUnique words in text A:")
    print(ftA)

    print("\nUnique words in text B:")
    print(ftB)

main()
```
