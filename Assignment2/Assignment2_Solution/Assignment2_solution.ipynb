# Majority-Class Prediction

```python
historical = DataTable(DATA_DIR / "cancer_historical1.csv")
new_cases = DataTable(DATA_DIR / "cancer_new_cases1.csv", historical)

tumor_types = historical.get_column("TumorType")

benign_count = tumor_types.count(0.0)
malignant_count = tumor_types.count(1.0)

mode_predictor2 = MajorityClassPredictor(historical)
patient = new_cases.get_row(0)
value = mode_predictor2.predict(patient)

print("Benign cases:", benign_count)
print("Malignant cases:", malignant_count)
print("Majority class:", mode_predictor2.majority_class)
print("expect return 0.0 for benign in patient: ", value)
```

# Nearest-Neighbout Prediction

```python
nearest_predictor = NearestNeighborPredictor(historical)

patient2 = new_cases.get_row(0)
closest_patient = None
smallest_distance = None

for historical_patient in historical.data:
    current_distance = nearest_predictor.distance(
        patient2, historical_patient
    )

    if (smallest_distance is None or
            current_distance < smallest_distance):
        smallest_distance = current_distance
        closest_patient = historical_patient

print("Distance to closest historical case:", smallest_distance)
print("Closest case tumor type:", closest_patient["TumorType"])
print("Nearest Neighbor prediction:",
      nearest_predictor.predict(patient2))
```

# Single OutOfSampleError

```python
historical2 = DataTable(DATA_DIR / "cancer_historical1.csv")
new_cases2 = DataTable(DATA_DIR2 / "test_oos_single.csv")

# Create predictors
mode_predictor = MajorityClassPredictor(historical2)
nn_predictor = NearestNeighborPredictor(historical2)

mode_predictions = []
nn_predictions = []


# Predict every new case
for i in range(len(new_cases2.data)):

    patient = new_cases2.get_row(i)
    

    try:
        mode_predictions.append(
            mode_predictor.predict(patient)
        )

        nn_predictions.append(
            nn_predictor.predict(patient)
        )

    except OutOfSampleError as e:
        print(e)
```

# Accuracy Calculation

```python
# Load datasets
historical = DataTable(DATA_DIR / "cancer_historical1.csv")
new_cases = DataTable(DATA_DIR / "cancer_new_cases1.csv", historical)
labels = DataTable(DATA_DIR / "cancer_new_cases_labels1.csv")

# Create predictors
mode_predictor = MajorityClassPredictor(historical)
nn_predictor = NearestNeighborPredictor(historical)

mode_predictions = []
nn_predictions = []

# Predict every new case
for i in range(len(new_cases.data)):

    patient = new_cases.get_row(i)

    try:
        mode_predictions.append(
            mode_predictor.predict(patient)
        )

        nn_predictions.append(
            nn_predictor.predict(patient)
        )

    except OutOfSampleError as e:
        print(e)

# Get true labels
actual_labels = labels.get_column("TumorType")

print("Historical rows:", len(historical.data))
print("New case rows:", len(new_cases.data))
print("Label rows:", len(labels.data))

# Accuracy
print(
    "Mode accuracy:",
    calculate_accuracy(mode_predictions, actual_labels)
)

print(
    "Nearest Neighbor accuracy:",
    calculate_accuracy(nn_predictions, actual_labels)
)
```
