# Task 1 - Load and Extract Data

```python
import numpy as np

stocks = np.genfromtxt('stock_prices_intro.csv', delimiter=',', names=True, dtype=None, encoding="utf-8")
print(stocks)
```

```python
date = stocks['Date']
openp = stocks['Open']
closep = stocks['Close']
vol = stocks['Volume']

print("Dates:", date)
print("Open Prices:", openp)
print("Close Prices:", closep)
print("Volume:", vol)
```

# Task 2 - Daily Rates

```python
# I was trying something. Doesn't effect the rest of the code from running

rate_list = []

for i in range(len(stocks)):
    rate = (closep[i]-openp[i])/openp[i]*100
    rate_list.append((date[i], (round(rate, 2))))

rate_arr = np.array(rate_list)
#print(rate_arr) #no formatting

for i in range(5): #print all (len(rate_arr))
    print(f"Date: {date[i]}  Rate: {rate_arr[i,1]}%") 
```

```python
rate = (closep-openp)/openp*100

for i in range(5): #5 because question asked for first 5 daily rates 
    print(f"Date: {date[i]}   Rate: {rate[i]:.2f}%") #:.2f rounds to 2 decimal places
```

# Task 3 - Basic Stats

```python
mean =np.mean(rate)
median =np.median(rate)
std =np.std(rate)
print(f"Mean: {mean:.2f}%")
print(f"Median: {median:.2f}%")
print(f"Standard Deviation: {std:.2f}%")
```

# Task 4 - Best and Worst Trading Days

```python
best_index=np.argmax(rate)
worst_index=np.argmin(rate)

print(f"Best Trading Day: {date[best_index]}, Opening Price: {openp[best_index]}, Closing Price: {closep[best_index]}, Daily Return Rate: {rate[best_index]:.2f}%")
```

# Task 5 - Up and Down Trading Days

```python
mask = rate > 0
mask_pos = np.sum(mask)

mask = rate < 0
mask_neg = np.sum(mask)

if mask_neg + mask_pos == len(rate):
    print(f'Number of days with positive returns: {mask_pos} days')
    print(f'Number of days with negative returns: {mask_neg} days')
else:
    print("error")
```

# Task 6 – Calculate Cumulative Return

```python
f = 1 + rate / 100
cum_return = (np.prod(f) - 1) * 100
print(f"Cumulative Return: {cum_return:.2f}%")
```
