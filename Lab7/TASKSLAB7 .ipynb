```python
import numpy as np 
#TASK 1 
# Load the CSV file 
data = np.genfromtxt("stock_prices_intro.csv",delimiter=",",names=True,
                     dtype=None,encoding="utf-8")

#Extract the four columns into separate arrays
date = data["Date"]
openp = data["Open"]
closep = data["Close"]
vol = data["Volume"]

print("Dates:", date)
print("Opening Prices:",openp)
print("Closing Prices:", closep)
print("Trading Volumes", vol)

# Test Task 1
print("\nTesting Task 1")

print("Number of records:", len(date))
print("Number of opening prices:", len(openp))
print("Number of closing prices:", len(closep))
print("Number of volumes:", len(vol))

assert len(date) == 20
assert len(openp) == 20
assert len(closep) == 20
assert len(vol) == 20

print("PASS: All four arrays contain 20 records")
```

```python
#TASK 2 

rate = (closep - openp) / openp * 100

print("\nTask 2 - First 5 Daily Returns")

for i in range(5):
    print(date[i], ":", round(rate[i], 2), "%")
```

```python
#TASK 3 

mean_rate = np.mean(rate)
median_rate = np.median(rate)
std_rate = np.std(rate)

print("\nTask 3 - Basic Stats")

print("Mean:", round(mean_rate, 2), "%")
print("Median:", round(median_rate, 2), "%")
print("Standard Deviation:", round(std_rate, 2), "%")
```

```python
# TASK 4 

#find highest and lowest 
best_day = np.argmax(rate)
worst_day = np.argmin(rate)


print("\nTask 4 - Best Trading Day")

print("Date:", date[best_day])
print("Opening Price:", openp[best_day])
print("Closing Price:", closep[best_day])
print("Daily Return:", round(rate[best_day], 2), "%")

print("\nTask 4 - Worst Trading Day")

print("Date:", date[worst_day])
print("Opening Price:", openp[worst_day])
print("Closing Price:", closep[worst_day])
print("Daily Return:", round(rate[worst_day], 2), "%")
```

```python
#TASK 5 

positive_days = rate > 0
negative_days = rate < 0

up_days = np.sum(positive_days)
down_days = np.sum(negative_days)

print("\nTask 5 - Up and Down Trading Days")

print("Positive trading days:", up_days)
print("Negative trading days:", down_days)
```

```python
#TASK 6 

f = 1 + rate / 100 

cum_return = (np.prod(f)-1) * 100

print("\nTask 6 - Cumulative Return")
print("Estimated Cumulative Return:", round(cum_return, 2), "%")
```
