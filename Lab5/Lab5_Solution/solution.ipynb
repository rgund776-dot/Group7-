```python
from characters import Fighter, Mage
from weapons import MeleeWeapon, RangedWeapon, MagicWeapon
from main import attack
```

# Normal case

```python
fighter = Fighter("Conan", 100, 20)
fighter_weapon = MeleeWeapon("Sword", 15)
mage = Mage("Gandalf", 80, 30)
mage_weapon = MagicWeapon("Staff", 25, 5, 10)

print("fighter: ",fighter)
print("fighter_weapon: ",fighter_weapon)
print("mage: ",mage)
print("mage_weapon: ",mage_weapon)
```

```python
attack(fighter, fighter_weapon, mage, 0)

print("fighter: ",fighter)
print("mage: ",mage)
```

```python
attack(mage, mage_weapon, fighter, 3)
print("fighter: ",fighter)
print("mage: ",mage)
```

# Error Case - Only Mage can use MagicWeapon

```python
attack(fighter, mage_weapon, mage, 0)
```

# Error Case - Only weapons that have a range attribute can be used at distances > 0

```python
attack(fighter, fighter_weapon, mage, 10)
```

# Error Case - Weapons that have a range cannot be used at distances above that range

```python
attack(mage, mage_weapon, fighter, 100)
```

# Special Case - fighter using ranged weapon 

```python
range_weapon = RangedWeapon("Bow", 10, 5)
ranged_fighter = Fighter("Archer", 90, 15)

attack(ranged_fighter, range_weapon, mage, 5)

print("ranged_fighter: ",ranged_fighter)
print("mage: ",mage)
```
